```python
min_frequency= 5 
```

```python
vocab = {
    "<PAD>" : 0,
    "<UNK>" : 1
}
```

```python
#now add the words that are min = 5 frequency 
for word , count in word_counts.items():
    if count >= min_frequency:
        vocab[word] = len(vocab)
print("vocabulary size:" , len(vocab))
print(list(vocab.items())[:20])
        
```

```python
#converting the words into ids 
def encode_review(review):
    tokens = review.split()
    ids = []
    for token in tokens:
        ids.append(vocab.get(token , vocab["<UNK>"]))
    return ids
    
```

```python
#testing it on one review
sample = train_df['clean_review'].iloc[0]
encoded = encode_review(sample)
print(sample.split()[:20])
print(encoded[:20])
```

trunacating and padding - becuase our revuews have different umber of tokens 
but pytorch needs rectangular tensor [batch_size , sequence length]
hence max_len = 500 if a review has more than 500 tokens truncate it 
if review has fewer tokens pad it 




```python
def encode_and_pad(review , max_len=500):
    ids = encode_review(review)
    if len(ids) > max_len:
        ids = ids[:max_len]
    else:
        ids = ids +[vocab["<PAD>"]] * (max_len - len(ids))
    return ids 
```

```python
#creating dataset 
from torch.utils.data import Dataset
class IMBDataset(Dataset):
    def __init__(self , dataframe):
        self.reviews = dataframe['clean_review'].tolist()
        self.labels = dataframe['label'].tolist()
    def __len__(self):
        return len(self.reviews)

    def __getitem__(self , index):
        review = encode_and_pad(self.reviews[index])
        label = self.labels[index]

        return torch.tensor(review , dtype = torch.long) , torch.tensor(label , dtype = torch.float32)



        
```

```python
#creating train and test datasets 

train_dataset = IMBDataset(train_df)
test_dataset = IMBDataset(test_df)
```

```python
#creating embedding 
import torch.nn as nn
embedding = nn.Embedding(
    num_embeddings = len(vocab),
    embedding_dim = 64,
    padding_idx = 0
)
embedded = embedding(x_batch)
print(x_batch.shape)
print(embedded.shape)

```

```python
# #mean pooling x.mean(dim = 1) dim=0 - reviews , dim=1 - tokens dim=2 embedding values 
# #hence we are supposed to average across the 500 tokens 
# # review_vector = embedded.mean(dim = 1)

# #rather than using if we use mask 
# embedding = 32 , 500 , 64 
# masking tells us what is true and padded vectors 
# example in a review 5 words are real remaining 495 are 0 padded then each 64 - dim = [64]...5 times 5*64 
# we want 1 * 64 we have different methods one is max pooling 

```

```python
# review_vector = embedded.mean(dim = 1)
# mask = (x_batch != 0)
mask = (x_batch != 0).unsqueeze(-1)
embedded = embedded * mask  #this will give 0 for padded and 1 for real vector it keeps it 
#calxulate the sum of real tokens 
sum_embeddings = embedded.sum(dim = 1)
#divide by real tokens 
num_tokens = mask.sum(dim = 1)
review_vector = sum_embeddings / num_tokens.clamp(min = 1)
#averaging only the real tokens hence 500 to ignoring padding - averaging real tokens - one 64 dimensional vector 
# [32 , 500 , 64] - masked mean pooling - [32 , 64]



```

Vector                 Mask       Result

[0.2, 0.5, -0.1]   ×   True   → [0.2, 0.5, -0.1]
[0.7, 0.1,  0.3]   ×   False  → [0.0, 0.0,  0.0]
This is also called masking by multiplication.

```python
# #building the actual model 
# input = [32 , 500] 
# embedding - [32 , 500 , 64]
# masked mean pooling [32 , 64]
# linear = 32 ,1 
# logit 
```

```python
#step 1 
import torch 
import torch.nn as nn
class SentimentalModel(nn.Module):
    def __init__(self , vocab_size , embedding_dim):
        super().__init__()
        self.embedding = nn.Embedding(
            num_embeddings = vocab_size,
            embedding_dim = embedding_dim ,
            padding_idx = 0
        ) 
        self.linear = nn.Linear(embedding_dim,1)
    def forward(self , x):
         embedded = self.embedding(x) #batch , 500
         mask = (x!= 0).unsqueeze(-1) #batch , 500 , 1
         embedded = embedded * mask #remove the pad Vectors
         sum_embeddings = embedded.sum(dim =1 ) #batch , 64

         num_tokens = mask.sum(dim =1) #number of real tokens 
         review_vector =(                                 #[batch , 64]
                sum_embeddings / num_tokens.clamp(min = 1)
                
            )
            #[batch , 1]

         logits = self.linear(review_vector)
         return logits

            
            


            
```

```python
#creatng a model 
#len(vocab) = 59847
#embedding_dim = 64 

model = SentimentalModel(vocab_size = len(vocab),
                        embedding_dim = 64)

```

```python
# $testing it 
logits = model(x_batch)
print(logits.shape)
```

```python
#the embedding and linear layer contain only randomly paramaterized 
#now we are supposed to train it 
#1 . LOSS function - binary classification - 0 , 1: BCEWithLogistsLOss
criterion = nn.BCEWithLogitsLoss()
logits = model(x_batch)
loss = criterion(
    logits.squeeze(1) , #squeeze(1) - we want 32 because y label = 32 
    y_batch
)
print(loss)
```

```python
#with training 
optimizer = torch.optim.Adam(
    model.parameters() , # now our model has learnable parameters 
    lr = 0.001
)
```

```python
#training step 
optimizer.zero_grad()
logits = model(x_batch)
loss = criterion(logits.squeeze(1),
                y_batch)
loss.backward()
optimizer.step()
print(loss.item())
```

40k training reviews (datset) = 32 batch size 
splited - batch 1 ---batch2---batch3 ----batch1250  40k/32 = 1250 batches 

```python
num_epochs = 3 
for epoch in range(num_epochs):
    model.train()

    total_loss = 0       # reset for this epoch

    for x_batch, y_batch in train_loader:

        optimizer.zero_grad()

        logits = model(x_batch)
#we pass the raw scores into loss not probabilities loogit are raw scores 
        loss = criterion(
            logits.squeeze(1),
            y_batch
        )  

        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    average_loss = total_loss / len(train_loader)

    print(f"Epoch {epoch + 1}:Loss = {average_loss:4f}")
```

```python
#model weights doesnt resets but loss resets at each epoch 
#loss can fluctuates between the epochs 
#gradient tells which direction increases the loss 
#example : if i increase this wieght , what happens to the loss ? 
```

```python
#evaluation mode 
model.eval()
with torch.no_grad():
    x_batch , y_batch = next(iter(test_loader))
    logits = model(x_batch).squeeze(1)
    probabilities = torch.sigmoid(logits)
    predictions = (probabilities >= 0.5).float()
print("actual labels :" , y_batch[:10])
print("probabilities:" , probabilities[:10])
print("predictions: " , predictions[:10])
```

```python
#to  
model.eval()
correct = 0
total = 0
all_predictions = []
all_labels = []
with torch.no_grad():
    for x_batch , y_batch in test_loader:
        logits = model(x_batch).squeeze(1)
        probabilities = torch.sigmoid(logits)
        predictions = (probabilities >= 0.5).float()
        correct += (predictions == y_batch).sum().item()
        total += y_batch.size(0)
        all_predictions.extend(predictions.cpu().numpy())
        all_labels.extend(y_batch.cpu().numpy())
accuracy = correct / total 
print(f"correct predcitions:{correct}/{total}")
print(f"test accuracy:{accuracy * 100:.2f}%")
#accuary is overall score not h=just about posistie 
# 45 + 44 = 89
#89 / 100 * 100 = 89 % 
```

```python
# confusion matrix- tells us which precitions are correct and which are wrong
from sklearn.metrics import confusion_matrix 
cm = confusion_matrix(all_labels , all_predictions)
print(cm)
```

```python
# preciosn  - of all the reviews model predicted as positive , how many were actually positive 
# precision = tp/tp+fp
# recall = true positive / true positive + false negative hence more false negatives more recalll is decreases 

#f1 score = combining precision + recall gives high score when r and p are high low when they are low 


#f1 = 2 * (p * r)/ (p + r)

from sklearn.metrics import f1_score 
f1 = f1_score(all_labels , all_predictions)
print(f"f1 score : {f1:.4f} ")
print(f"f1 score:{f1 * 100:.2f}%")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

print("Accuracy:", accuracy_score(all_labels, all_predictions))
print("Precision:", precision_score(all_labels, all_predictions))
print("Recall:", recall_score(all_labels, all_predictions))
print("F1-score:", f1_score(all_labels, all_predictions))

print(classification_report(
    all_labels,
    all_predictions,
    target_names=["Negative", "Positive"]
))
```
